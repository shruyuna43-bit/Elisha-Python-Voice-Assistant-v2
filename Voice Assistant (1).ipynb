```python
import pyttsx3
import speech_recognition as sr
import webbrowser


# Initialize recognizer
recognizer = sr.Recognizer()


def speak(text):
    print("Elisha:", text)

    engine = pyttsx3.init()
    engine.setProperty("rate", 175)
    engine.setProperty("volume", 1.0)

    engine.say(text)
    engine.runAndWait()

    engine.stop()






def listen():
    try:
        with sr.Microphone() as source:
            print("\nAssistant is listening...")

            recognizer.adjust_for_ambient_noise(source, duration=0.5)

            audio = recognizer.listen(
                source,
                timeout=5,
                phrase_time_limit=8
            )

            print("Recognizing...")

            command = recognizer.recognize_google(audio)

            command = command.lower().strip()

            print("You said:", command)

            return command

    except sr.WaitTimeoutError:
        print("No speech detected.")
        return ""

    except sr.UnknownValueError:
        speak("Sorry, I couldn't understand that.")
        return ""

    except sr.RequestError:
        speak("Speech recognition service is unavailable.")
        return ""

    except OSError:
        speak("Microphone not found.")
        return ""


def run():

    speak("Hello! I'm your voice assistant. How can I help you?")

    while True:

        command = listen()

        if not command:
            continue

        if "hello" in command or "hie" in command:
            speak("Hello! How can I help you?")

        elif "your name" in command:
            speak("My name is Elisha.")
        elif "google" in command:
            speak("I am openinng google")
            webbrowser.open("https://www.google.com/")
        elif "youtube" in command:
            speak("I am openinng Youtube")
            webbrowser.open("https://www.youtube.com/")
        
        elif "stop" in command or "goodbye" in command:
            speak("Goodbye! Have a nice day.")
            break

        else:
            speak("I can't recognize your command!")

run()
```
